# Notebook to remove .0_0000 from the end

```python
import os
import re

############### Change this to the folder containing your files ###############
folder = "data/STO_Substrate_06102026"
###############################################################################

for filename in os.listdir(folder):
    ext = os.path.splitext(filename)[1]
    name = os.path.splitext(filename)[0]

    # Find final number
    match = re.search(r"(\d+)$", name)

    if match:
        number = int(match.group(1))

        # Get original prefix
        prefix = name[:match.start()]

        # Remove ".0_" from the end
        prefix = re.sub(r"\.0_$", "", prefix)

        new_name = f"{prefix}_{number}{ext}"

        old_path = os.path.join(folder, filename)
        new_path = os.path.join(folder, new_name)

        if filename != new_name:
            os.rename(old_path, new_path)
            print(f"{filename} → {new_name}")
```

## Fix name_ where to many underscores

```python
############### Change this to the folder containing your files ###############
folder = "data/STO_Substrate_06102026"
###############################################################################

for filename in os.listdir(folder):
    name, ext = os.path.splitext(filename)

    # Replace double underscore before the final number with one underscore
    new_name = re.sub(r"__(\d+)$", r"_\1", name) + ext

    if filename != new_name:
        old_path = os.path.join(folder, filename)
        new_path = os.path.join(folder, new_name)

        os.rename(old_path, new_path)
        print(f"{filename} → {new_name}")
```
