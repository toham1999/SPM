# Notebook to add ROI to AFM

## Setup
Packages and images

### Packages

```python
import matplotlib.pyplot as plt
from matplotlib.patches import Rectangle
import os
```

### Import images

### Image 1

```python
filename = "SPM/data/2026-10-06_STO111_Substrate/images/STO111_Substrate_0.png"

img = plt.imread(filename)

fig, ax = plt.subplots(figsize=(7, 6))

ax.imshow(img)

# ROI: x, y, width, height in pixels
x, y = 160, 145
w, h = 175, 175

ax.add_patch(Rectangle((x, y), w, h,fill=False,edgecolor="white",linewidth=2))

ax.text(x, y - 10, "ROI", color="white", fontsize=14)

ax.axis("off")
plt.tight_layout()
name, ext = os.path.splitext(filename)
plt.savefig(f"{os.path.dirname(name)}/{os.path.basename(name)}_ROI{ext}",transparent=True,dpi=300,bbox_inches="tight")
plt.show()


show_roi_border = False

if show_roi_border:
    # Display ROI with surrounding border
    border = 30

    x0 = max(0, x - border)
    y0 = max(0, y - border)
    x1 = min(img.shape[1], x + w + border)
    y1 = min(img.shape[0], y + h + border)

    crop = img[y0:y1, x0:x1]

    fig, ax = plt.subplots(figsize=(6, 6))
    ax.imshow(crop)

    ax.add_patch(Rectangle((x - x0, y - y0),w, h,fill=False,edgecolor="white",linewidth=2))

    ax.axis("off")
    plt.tight_layout()
    plt.show()
```
